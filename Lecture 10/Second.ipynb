```python
with open('input.txt','r') as f:
    text=f.read()
data=text[:1000]
```

```python
import tiktoken
enc=tiktoken.get_encoding('gpt2')
tokens=enc.encode(data)
print(tokens[:24])
```

```python
import torch
buf=torch.tensor(tokens[:24+1])
x=buf[:-1].view(4,6)
y=buf[1:].view(4,6)
print(x)
print(y)
```
